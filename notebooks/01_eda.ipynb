# Exploratory Data Analysis (EDA)
## MovieLens 20M Dataset

In this notebook, we analyze the MovieLens 20M dataset to understand the distribution of ratings, genres, and user activity.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")
%matplotlib inline
```

### 1. Load Data
We load the processed Parquet files for efficiency.

```python
DATA_PATH = os.path.join("..", "data", "processed")

movies = pd.read_parquet(os.path.join(DATA_PATH, "movies.parquet"))
ratings = pd.read_parquet(os.path.join(DATA_PATH, "ratings.parquet"))

print(f"Movies shape: {movies.shape}")
print(f"Ratings shape: {ratings.shape}")
```

### 2. Ratings Distribution

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='rating', data=ratings, palette='viridis')
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

### 3. Most Popular Movies

```python
popular_movies = ratings.groupby('movieId').size().reset_index(name='count')
popular_movies = popular_movies.merge(movies, on='movieId')
popular_movies = popular_movies.sort_values('count', ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x='count', y='title', data=popular_movies, palette='magma')
plt.title('Top 10 Most Rated Movies')
plt.xlabel('Number of Ratings')
plt.ylabel('Movie Title')
plt.show()
```

### 4. Genre Analysis

```python
genres_exploded = movies.explode('genres')
genre_counts = genres_exploded['genres'].value_counts().reset_index()
genre_counts.columns = ['genre', 'count']

plt.figure(figsize=(12, 8))
sns.barplot(x='count', y='genre', data=genre_counts, palette='coolwarm')
plt.title('Number of Movies per Genre')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```
